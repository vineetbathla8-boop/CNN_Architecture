```python
import tensorflow as tf
from tensorflow.keras.applications import VGG16,VGG19
from tensorflow.keras.applications.vgg16 import preprocess_input, decode_predictions
from tensorflow.keras.preprocessing.image import load_img, img_to_array
import numpy as np
```

```python
# loading the predefined model
model = VGG16(weights="imagenet")
```

```python
model_19 = VGG19(weights = "imagenet")
```

```python
# creating the prediction function

def prediction_image(model,img_path):

  img = load_img(img_path,target_size = (224,224))

  img_array = img_to_array(img)

  img_array = np.expand_dims(img_array,axis = 0)

  img_array = preprocess_input(img_array)

  pred = model.predict(img_array)

  label = decode_predictions(pred,top = 5)
  return label[0]
```

```python
sample_img_path = "/content/sample_data/vulture.webp"

prediction = prediction_image(model,sample_img_path)

for i,(imagnet_id,labels,score) in enumerate(prediction):
  print(f"{i+1}. {labels} : {score*100:.4f}%")
```
